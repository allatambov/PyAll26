### Активировать тесты

Запустите следующую ячейку, чтобы иметь возможность запускать тесты.

```python
# Фабрика тестов для проверки программ, принимающих данные через input()

from collections import deque

class Tester(object):
    def __init__(self, inp):
        self.outputs = []
        self.inputs = deque(inp)
    def print(self, *args, sep = " ", end = "\n"):
        text = sep.join(map(str, args)) + end
        newlines = text.splitlines(keepends=True)
        if self.outputs and self.outputs[-1] and self.outputs[-1][-1] != "\n" and newlines:
            self.outputs[-1] += newlines[0]
            self.outputs.extend(newlines[1:])
        else:
            self.outputs.extend(newlines)
            
    def input(self, *args):
        assert self.inputs, "Вы пытаетесь считать больше элементов, чем предусмотрено условием"
        return self.inputs.popleft()
    def __enter__(self):
        global print
        global input
        print = self.print
        input = self.input
        return self.outputs
    def __exit__(self, *args):
        global print
        global input
        del print
        del input
```

Напишите программу, которая запрашивает у пользователя название текущей станции метро и выводит на экран сообщение вида:

    Следующая станция: [станция].

Считайте, что пользователь знает, какая станция конечная, и не будет запрашивать станцию, следующую после Новокосино. Более общий вариант задачи – см. **задачу №6**.

```python
def metro():
    
    line = ["Третьяковская", "Марксистская", "Площадь Ильича", "Авиамоторная", "Шоссе Энтузиастов",
    "Перово", "Новогиреево", "Новокосино"]

    # находим индекс текущей станции и увеличиваем его на 1,
    # выбираем нужный элемент из line и подставляем в шаблон
    
    station = input()
    i = line.index(station) + 1
    print(f"Следующая станция: {line[i]}.")

metro()
```

### Задача 2. В траве сидел кузнечик...

В мультфильме «Незнайка на Луне» (1997) Незнайка, находясь в ракете, передразнивая компьютер, поёт:

    В траве сидел компьютер, 
    Зелёненький он был,
    И всё он перепутал,
    И всё он позабыл!
    
Напишите программу, которая запрашивает с клавиатуры строку с текстом (весь текст маленькими буквами) и заменяет в ней слово «кузнечик» на «компьютер», а слово «компьютер» – на «кузнечик». 

Уточнения:

* Считайте, что слова «кузнечик» и «компьютер» не встречаются в строке с текстом одновременно.
* Слова во множественном числе и других падежах не рассматриваем, более гибкая замена – это отдельная более сложная задача. 


**Пример работы программы**

Ввод:

    компьютер завис 

Вывод:

    кузнечик завис
    
Ввод:

    кузнечик – насекомое?

Вывод:

    компьютер – насекомое?
    
Ввод:

    один кузнечик и второй кузнечик 

Вывод:

    один компьютер и второй компьютер

```python
def grasshopper():

    # проверяем наличие слова и если есть, заменяем,
    # если убрать проверку условий, две последовательные замены 
    # вернут строку в исходное состояние
    
    text = input()
    if "кузнечик" in text:
        new = text.replace("кузнечик", "компьютер")
    if "компьютер" in text:
        new = text.replace("компьютер", "кузнечик")
    print(new)

grasshopper()
```

### Задача 3. Count on me

Напишите программу, которая:

* запрашивает у пользователя с клавиатуры строку с текстом
* с новой строки запрашивает символ (любой символ, буква, цифра, знак препинания, пробел)

и выводит на экран число вхождений этого символа во введённую строку с текстом. 

**Важно.** Если символ является буквой, число вхождений нужно считать для этой буквы как в нижнем, так и верхнем регистре. Например, в тексте `Text` буква `t` должна быть найдена два раза, так как здесь есть и заглавная, и строчная буква `t`.

**Пример работы программы**

Ввод:

    text
    t

Вывод:

    2
    
Ввод:

    Some Interesting Text Here
    t
    
Вывод:

    4
    
Ввод:

    Some Interesting Text Here
    T
    
Вывод:

    4
    
Ввод:

    allatambov@gmail.com
    @
    
Вывод:

    1
    
Ввод:

    Слова, слова, слова
    ,
    
Вывод:

    2

```python
def count_on_me():

    # самый простой способ – сразу перевести текст в нижний регистр
    # и символ тоже, а потом просто посчитать
    
    text = input().lower()
    symbol = input().lower()
    print(text.count(symbol))

count_on_me()
```

```python
test_data = [
    ("text;t", "2"),
    ("Some Interesting Text Here;t", "4"),
    ("Some Interesting Text Here;T", "4"),
    ("allatambov@gmail.com;@", "1"),
    ("Слова, слова, слова;,", "2"),
    ("Без умолку безумная девица...;б", "2"),
    ("Кричала: «Ясно вижу Трою, павшей в прах!»;«", "1"),
    ("Но ясновидцев — впрочем, как и очевидцев —;е", "4"),
    ("Во все века сжигали люди на кострах.;В", "3"),
    ("Владимир Высоцкий;И", "3"),
    ("Песня о вещей Кассандре 1967;1", "1"),
    ("Песня о вещей Кассандре 1967; ", "4")
]

for inp, out in test_data:
    with Tester(inp.split(";")) as t:
        count_on_me()
        assert len(t) == 1, "Вам нужно вывести ровно одну строку с ответом"
        assert t[0] == out+"\n", "Неверный ответ, была введена строка " + inp
```

### Задача 4. Разделяй и властвуй

Напишите программу, которая запрашивает с клавиатуры текст (слова через пробел без знаков препинания) и выводит на экран список слов в верхнем регистре (заглавными буквами), отсортированный по алфавиту. 

```python
def words_count():

    # чтобы не переводить слова в списке к верхнему регистру
    # через перебор в цикле, до split() переводим все в большие буквы
    
    words = input().upper().split()
    words.sort()
    print(words)
    
words_count()
```

```python
test_data = [
    ("я с памятной доски сотру все знаки", "['ВСЕ', 'ДОСКИ', 'ЗНАКИ', 'ПАМЯТНОЙ', 'С', 'СОТРУ', 'Я']"),
    ("где грифель мой я это запишу", "['ГДЕ', 'ГРИФЕЛЬ', 'ЗАПИШУ', 'МОЙ', 'ЭТО', 'Я']"),
    ("прощай прощай и помни обо мне", "['И', 'МНЕ', 'ОБО', 'ПОМНИ', 'ПРОЩАЙ', 'ПРОЩАЙ']"),
    ("there are more things in heaven and earth Horatio", "['AND', 'ARE', 'EARTH', 'HEAVEN', 'HORATIO', 'IN', 'MORE', 'THERE', 'THINGS']"),
    ("Than are dreamt of in your philosophy", "['ARE', 'DREAMT', 'IN', 'OF', 'PHILOSOPHY', 'THAN', 'YOUR']")
]

for inp, out in test_data:
    with Tester([inp]) as t:
        words_count()
        assert len(t) == 1, "Вам нужно вывести ровно одну строку с ответом"
        assert t[0] == out+"\n", "Неверный ответ, была введена строка " + inp
```

В списке `old` сохранены названия файлов с описанием или изображением планет Солнечной системы. Напишите программу, которая сохраняет в список `new` только названия файлов с изображениями – файлы с расширениями `.png` и `.jpeg`. Элементы в списке `new` должны быть отсортированы по алфавиту. Если в списке `old` нет файлов с расширениями `.png` и `.jpeg`, в `new` должен сохраняться пустой список.

**Важно.** Для корректной проверки кода итоговый список должен называться именно `new`. Не стирайте в коде ниже строку с `return new`.

**Пример работы программы**

Список на входе:

    ["Jupiter.pdf", "Jupiter.jpeg", "Mars.png", "Earth.doc", 
    "Uranus.docx", "Saturn.jpeg", "Earth.png"]

Список на выходе:

    ['Earth.png', 'Jupiter.jpeg', 'Mars.png', 'Saturn.jpeg']

Список на входе:

    ["Mars.pdf", "Earth.jpeg", "Venus.pdf", "Mercury.jpeg", "Jupiter.docx"]

Список на выходе:

    ['Earth.jpeg', 'Mercury.jpeg']
    
Список на входе:

    ["Venus.pdf", "Jupiter.docx"]

Список на выходе:

    []

```python
# примеры списка old, на каждом из них 
# можно проверить работу своего кода

old = ["Jupiter.pdf", "Jupiter.jpeg", "Mars.png", 
       "Earth.doc", "Uranus.docx", "Saturn.jpeg", 
       "Earth.png"]

old = ["Venus.docx", "Mars.pdf", "Earth.jpeg", 
      "Venus.png", "Mercury.jpeg", "Jupiter.docx"]
```

```python
def hello_jupiter(old):

    new = []
    for file in old:
        if file.endswith(".png") or file.endswith(".jpeg"):
            new.append(file)
    new.sort()
    
    return new
```

```python
# более изящный вариант – метод .endswith() может принимать сразу
# несколько строк на вход в виде кортежа, or можно не использовать

def hello_jupiter(old):

    new = []
    for file in old:
        if file.endswith((".png", ".jpeg")):
            new.append(file)
    new.sort()
    
    return new
```

```python
assert hello_jupiter(["Jupiter.pdf", "Jupiter.jpeg", "Mars.png", 
                      "Earth.doc", "Uranus.docx", "Saturn.jpeg", 
                      "Earth.png"]) == ['Earth.png', 'Jupiter.jpeg', 
                                            'Mars.png', 'Saturn.jpeg']
assert hello_jupiter(["Venus.docx", "Mars.pdf", "Earth.jpeg", 
                      "Venus.png", "Mercury.jpeg", 
                      "Jupiter.docx"]) == ['Earth.jpeg', 'Mercury.jpeg', 
                                           'Venus.png']
assert hello_jupiter(["Mars.pdf", "Earth.jpeg", "Venus.pdf", 
               "Mercury.jpeg", "Jupiter.docx"]) == ['Earth.jpeg', 'Mercury.jpeg']

assert hello_jupiter(["Venus.pdf", "Jupiter.docx"]) == []
```

### Задача 7*. Зелёненький он был...

*Более общая версия задачи №2*

В мультфильме «Незнайка на Луне» (1997) Незнайка, находясь в ракете, передразнивая компьютер, поёт:

    В траве сидел компьютер, 
    Зелёненький он был,
    И всё он перепутал,
    И всё он позабыл!
    
Напишите программу, которая запрашивает с клавиатуры строку с текстом (весь текст маленькими буквами) и заменяет в ней слово «кузнечик» на «компьютер», а слово «компьютер» – на «кузнечик». Считайте, что слова «кузнечик» и «компьютер» **могут встречаться** в строке с текстом одновременно.

Слова во множественном числе и других падежах не рассматриваем, более гибкая замена – это отдельная более сложная задача. 


**Пример работы программы**

Ввод:

    компьютер завис 

Вывод:

    кузнечик завис
    
Ввод:

    кузнечик – насекомое?

Вывод:

    компьютер – насекомое?
    
Ввод:

    кузнечик прыгает, компьютер работает

Вывод:

    компьютер прыгает, кузнечик работает

```python
def grasshopper2():
    
    # ну тут не то что бы задача со звездочкой – просто 
    # догадаться заменить нужное слово на третье слово, которого 
    # точно в тексте быть не может

    text = input()
    text = text.replace("компьютер", "$$$")
    text = text.replace("кузнечик", "компьютер")
    text = text.replace("$$$", "кузнечик")
    print(text)

grasshopper2()
```

```python
test_data = [
    ("компьютер завис", "кузнечик завис"),
    ("кузнечик кузя", "компьютер кузя"),
    ("один кузнечик и второй кузнечик", 
     "один компьютер и второй компьютер"),
    ("компьютер перегрелся", "кузнечик перегрелся"),
    ("программист перегрелся, кузнечик запрыгал",
     "программист перегрелся, компьютер запрыгал"),
    ("в траве сидел кузнечик", "в траве сидел компьютер"),
    ("в сиде травел кузнечик", "в сиде травел компьютер"),
    ("кузнечик прыгает, компьютер работает", "компьютер прыгает, кузнечик работает"),
    ("компьютер и кузнечик", "кузнечик и компьютер"),
    ("кузнечик кузя кузнечик тузя компьютер lenovo", "компьютер кузя компьютер тузя кузнечик lenovo"), 
]

for inp, out in test_data:
    with Tester([inp]) as t:
        grasshopper2()
        assert len(t) == 1, "Вам нужно вывести ровно одну строку с ответом"
        assert t[0] == out+"\n", "Неверный ответ, была введена строка " + inp
```
