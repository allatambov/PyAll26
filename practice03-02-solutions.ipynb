### Немного теории: чтение и запись текстовых файлов

Для работы с txt-файлами (неважно, хотим ли мы считывать содержимое файла или, наоборот, записывать туда что-то новое), нам понадобится функция `open()`. У этой функции есть разные аргументы, основные такие:

* аргумент `file`: название файла или путь к нему в виде обычной строки;
* аргумент `mode`: режим работы с файлом;
* аргумент `encoding`:  кодировка файла.

**Режим работы с файлом.** Есть три основных режима: `'r'` (от *read*) – открыть файл только для чтения; `'a'` (от *append*) – открыть файл для добавления к нему строк; `'w'` (от *write*) – открыть файл для записи с нуля, то есть создать новый пустой текстовый файл.

*Важный момент 1.* Если файл с некоторым названием уже существует, при создании нового файла с таким названием в режиме `'w'` старый файл молча сотрется и заменится новым. Существует режим `'x'`, он позволяет создать новый файл, но если файл с таким названием уже существует, будет возвращена ошибка (более безопасный вариант *write*).

*Важный момент 2.* Если при открытии файла не указать режим, по умолчанию он будет открыт только для чтения.

**Кодировка.** По умолчанию на Unix-системах (Linux, Mac) используется кодировка UTF-8, на Windows возможны варианты, чаще всего кодировка CP-1251, по крайней мере, для файлов на кириллице. Если файл был создан на Mac/Linux, и он содержит текст не на английском (точнее, не на простой латинице, на языках, где есть диакритические знаки вроде ударений или умляутов), при открытии такого файла на Windows без указания кодировки будут проблемы. В обратную сторону это тоже работает: файл, созданный на Windows, не будет корректно считываться на Mac/Linux. Поэтому кодировку лучше всегда указывать или, по крайней мере, знать, что если Python выдает ошибку с сообщением о том, что он не может что-то декодировать, то аргумент с кодировкой надо добавить.

*Важный момент 3.* Функция `open()` работает только с локальными файлами, то есть с теми, что есть на компьютере. Находить и открывать файл по ссылке она не умеет.

Создадим файл `test.txt` – откроем его **для записи с нуля** с флагом `"w"` (откроем соединение с файлом):

```python
myfile = open("test.txt", "w")
print(myfile)
print(type(myfile))
```

> Это объект типа `_io.TextIOWrapper`, который представляет собой открытое соединение с файлом. К этому соединению можно применять методы для чтения или записи текста.

Добавим в него записи – строки с текстом:

```python
# метод .write() записывает строку с текстом в файл,
# не добавляя при этом отступы и не переходя на новую строку,
# поэтому здесь мы сами добавили \n в нужных местах

myfile.write("THIS IS A TEST\nWAIT\n")
```

> По идее, теперь в файле должны быть две строки: `THIS IS A TEST` и `WAIT`. Но если мы откроем файл, текста там не окажется. Почему? Изменения нужно сохранить, закрыв файл!

Закроем файл (закроем соединение с файлом) – сохраним изменения:

```python
myfile.close()
```

Если не хочется каждый раз писать `.close()` по окончании работы, можно использовать конструкцию `with ... as ...`. Чаще всего эта конструкция и используется на практике:

```python
with open("test.txt", "w") as myfile:
    myfile.write("THIS IS A TEST\nWAIT\n")
print("Получилось то же самое!")
```

> Здесь строка `with open("test.txt", "w") as myfile` эквивалентна записи `myfile = open("test.txt", "w")`, а выход из тела конструкции (когда блок кода с отступами заканчивается) эквивалентен записи `myfile.close()`.

Откроем файл **в режиме чтения и дозаписи** (флаг `"r+"`). После применения метода `.read()`, файл будет прочитан с самого начала, и мы увидим текст, сохраненный в файле:

```python
with open("test.txt", "r+") as myfile:
    print(myfile.read())
```

Благодаря `.read()` курсор, который «путешествует» по файлу, остановился после последней строки с текстом. А значит, мы можем дописать что-то в конец файла и сохранить изменения:

```python
# метод .writelines() записывает в файл список строк,
# если на входе .writelines() список не из текстовых значений, 
# получим ошибку

with open("test.txt", "r+") as myfile:
    print(myfile.read())
    myfile.writelines(["3\n", "2\n", "1\n", "Start!\n"])
```

Если мы не напишем строку `myfile.read()`, Python предварительно не считает содержимое файла, а значит, курсор, «путешествующий» по файлу, будет стоять не в конце последней строки, а в начале первой. И тогда мы просто запишем новый текст поверх старого:

```python
# для чистоты эксперимента повторим создание файла,
# чтобы файл бы в том же состоянии, что и перед 
# запуском предыдущей ячейки

with open("test.txt", "w") as myfile:
    myfile.write("THIS IS A TEST\nWAIT\n")
```

```python
# запустим и откроем файл вручную

with open("test.txt", "r+") as myfile:
    myfile.writelines(["3\n", "2\n", "1\n", "Start!\n"])
```

Если нам не хочется «пролистывать» файл перед добавлением в него нового текста, мы можем открыть его в **режиме дозаписи и чтения** (флаг `"a+"`):

```python
# для чистоты эксперимента повторим создание файла

with open("test.txt", "w") as myfile:
    myfile.write("THIS IS A TEST\nWAIT\n")
```

```python
# запустим и откроем файл вручную

with open("test.txt", "a+") as myfile:
    myfile.writelines(["3\n", "2\n", "1\n", "Start!\n"])
```

**Резюме.** Кратко о режимах и методах:

* Если нужно открыть файл только для чтения, используйте флаг `r` или не указывайте ничего, этот флаг выставляется по умолчанию. Если файл открыт в режиме чтения, к соединению с ним нельзя будет применять методы `.write()` и  `.writelines()`. Поскольку дозапись запрещена, применение методов выдаст ошибку `UnsupportedOperation: not writable`. Зато можно применять методы `.read()` и `.readlines()`. Первый метод прочитывает файл целиком и возвращает весь текст в виде одной большой строки, второй – считывает файл построчно и возвращает список строк  с текстом.

* Если нужно создать новый файл и записать в него текст, используйте флаг `w`. Если файл открыт в режиме записи, к соединению можно применять методы `.write()` и  `.writelines()`. Первый принимает на вход строку с текстом, второй – список строк. Однако ни первый, ни второй метод не добавляет переход на новую строку, их надо добавлять самостоятельно. Например, если мы захотим записать числа 1, 2, 3, каждое с новой строки, в файл, созданный через `file = open("test.txt", "w")`, потребуется запустить код `file.write("1\n2\n3\n")` или `file.writelines(["1\n", "2\n", "3\n"])`. Последний способ хорош, если у нас есть готовый список строк для сохранения – к каждой из них предварительно можно доклеить `"\n"` в цикле, чтобы гарантировать перенос строк в файле.

* Если нужно к уже существующему файлу с текстом дописать строки, не порушив при этом его содержимое, используйте флаг `a`. В таком режиме не будут доступны методы `.read()` и `.readlines()`, поскольку файл открыт только для дозаписи. Можно применить методы `.write()` или `.writelines()`, они гарантированно допишут строки в конец файла.

* Если файл нужно и читать, и дозаписывать, потребуется смешанный режим работы: `a+` (дозапись + чтение) или `r+` (чтение + дозапись). Если используем `a+`, можно сразу применять `.write()` и `.writelines()` для дозаписи новых строк в конец, а потом применять `.read()` или `.readlines()` для считывания полученного текста. Если используем `r+`, то *перед* дозаписью текста через `.write()` и `.writelines()` нужно применить метод `.read()` или `.readlines()` для того, чтобы считать содержимое и перейти в конец файла. В противном случай файл будет обрабатываться с самого начала, и новые строки будут дописаны не в конец, а «поверх» уже имеющегося в файле текста. 

### Практика

Откроем файл `1.txt` в режиме чтения (флаг `"r"`, но можно не ставить, он используется по умолчанию):

```python
file = open("1.txt", encoding = "utf-8")
```

Раскомментируйте строки ниже и сравните результаты:

```python
print(file.read())
```

```python
print(file.readlines())
```

> Строка кода с `.read()` вернула текст целиком – одну большую строку с текстом внутри. Строка кода с `.readlines()` должна была вернуть список строк с текстом, но вернула пустой список строк. Это произошло из-за того, что после применения `.read()` выше, курсор, путешествующий по файлу, остановился в конце последней строки, а дальше в файле ничего нет. Поэтому откроем файл заново и применим только `.readlines()`:

```python
# lines – обычный список, один элемент = одна строка в файле
# всего строк 233

file = open("1.txt", encoding = "utf-8")
lines = file.readlines()
print(type(lines), len(lines))
```

### Задача 1

Напишите код, который сохранит в список `numbers` индексы строк в файле `1.txt`, содержащих название группы. Подумайте, как, используя список `numbers`, выбрать название и текст первой песни.

Подсказка: воспользуйтесь функцией `enumerate()`.

```python
numbers = []
for i, line in enumerate(lines):
    if line.startswith("The Beatles"):
        numbers.append(i)
print(numbers)
```

```python
# первая песня – строки от 0 до 13,
# вторая песня – строки от 13 до 37

song1 = lines[0:13]
song2 = lines[13:37]
print(song1)
print(song2)
```

```python
# при таком подходе мы сможем составить пары из последовательных элементов
# в numbers и подставлять их в срез как начало и конец,
# однако, чтобы не потерять последнюю песню, надо добавить в конец numbers
# индекс последней строки + 1, то есть длину списка lines

numbers.append(len(lines))
print(numbers)
```

### Задача 2

Функция `pairwise()` из модуля `itertools` создает пары последовательных элементов на основе последовательности:

```python
# примеры

print(list(itertools.pairwise(["Anna", "Bob", "Charlie", "Dick"])))
print(list(itertools.pairwise("abcde")))
```

Получите список таких пар на основе списка `numbers` и сохраните его в переменную `pairs`. Напишите цикл, который создаст список `songs` со строками из песен (один элемент `songs` – список из строк). 

```python
pairs = list(itertools.pairwise(numbers))
print(pairs)
```

**Важно.** Если у вас более старая версия Python, где в `itertools` отсутствует `pairwise()`, запустите код ниже для получения списка `pairs`:

```python
# на каждом шаге zip() склеивает элементы, убирая первый

pairs = list(zip(numbers, numbers[1:]))
print(pairs)
```

```python
# добавляем

songs = []
for p in pairs:
    song = lines[p[0]:p[1]]
    songs.append(song)

# проверяем – смотрим на первые два элемента

print(songs[0])
print(songs[1])
```

> Код ниже создает папку с названием *The Beatles* (функция `mkdir()` – от *make directory*), а затем делает эту папку рабочей (функция `chdir()` – от *change directory*).

```python
os.mkdir("The Beatles")
os.chdir("The Beatles")
print(os.getcwd())
```

### Задача 3

Напишите цикл, который создаст в рабочей папке (а после запуска кода выше мы в ней) текстовые файлы вида `The Beatles, Yesterday.txt` с текстами песен.

```python
# song[0] – забираем первую строку в списке song (название песни)
# song[0].strip() – убираем лишние пробелы – избавляемся от \n,
# song[0].strip() + ".txt" – доклеиваем расширение файла
# через open() создаем новый файл,
# в него записываем строки из song,
# закрываем для сохранения изменений

for song in songs:
    name = song[0].strip() + ".txt"
    new = open(name, "w")
    new.writelines(song)
    new.close()
```

### Задача 4

Напишите цикл, который переименует файлы в папке, убрав `The Beatles,` в начале, например, вместо `The Beatles, Yesterday.txt` должно остаться `Yesterday.txt`.

Подсказка: функция `os.rename()`, на первом месте указывается старое название, а на втором – новое.

**Совет.** Так как при работе в Jupyter Notebook могут сохраняться скрытые файлы, после получения перечня файлов в текущей папке, выберите для дальнейшей работы только файлы, оканчивающиеся на `.txt`.

```python
# names – текущие файлы в рабочей папке The Beatles

names = os.listdir()
print(names)
```

```python
# создаем список новых названий на основе старых в names

# проверяем, что файл текстовый – оканчивается на .txt,
# если да, название добавляем в old, затем получаем новое
# name.split(",") – возвращает список вида ['The Beatles', 'Yesterday\n']
# name.split(",")[1] – возвращает второй элемент вида 'Yesterday\n'
# name.split(",")[1].strip() – убирает отступ и получает элемент вида 'Yesterday'
# name.split(",")[1].strip() + ".txt" – получает элемент вида 'Yesterday.txt'

old = []
new = []

for name in names:
    if name.endswith(".txt"):
        old.append(name)
        new_name = name.split(",")[1].strip() + ".txt"
        new.append(new_name)
print(new)
```

```python
# перебираем пары старое название - новое название, 
# переименовываем через rename

for one, two in zip(old, new):
    os.rename(one, two)
```
