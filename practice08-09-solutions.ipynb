### Задача 2

Не создавая новых столбцов, вычислите, сколько команд получило золотых медалей больше, чем серебряных и бронзовых вместе взятых.

### Задача 3

Выведите таблицу частот для типа Игр (`games_type`). Посмотрите на значения и добавьте в `games`:

* бинарный столбец `summer`, где 1 соответствует летним Играм, а 0 – зимним;
* столбец `season`, где 1 соответствует летним играм, а 2 – зимним.

```python
games["games_type"].value_counts()
```

```python
# создаем булев столбец из True/False,
# переводим его в тип integer,
# чтобы True перешли в 1, False – в 0

games["summer"] = (games["games_type"] == "Summer").astype(int)
games["summer"].value_counts() # все хорошо
```

```python
# вариант 1: длинный
# пишем функцию, применяем в .apply()

def encode_season(x):
    if x == "Summer":
        y = 1
    else:
        y = 2
    return y

games["season"] = games["games_type"].apply(encode_season)
```

```python
# вариант 2: короткий
# пишем функцию через lambda сразу в .apply()

games["season"] = games["games_type"].apply(lambda x: 1 if x == "Summer" else 2)
```

### Задача 4

Добавьте в `games` бинарный столбец `matching`, где 1 соответствует наблюдениям, где выступающая страна совпадает с принимающей страной, а 0 – всем остальным наблюдениям. Другими словами, значения 1 должны стоять в тех случаях, когда команда играла «дома». Вычислите количество и долю таких случаев.

```python
games["matching"] = (games["country"] == games["host_country"]).astype(int)
```

```python
# количество и доля
print(games["matching"].sum())
print(games["matching"].mean())
```

### Задача 3

Сохраните строки, соответствующие звездам спектрального класса `O` и `M` , в датафрейм `stars_om`. Вычислите среднюю температуру таких звезд и округлите ее до второго знака после запятой.

```python
# смотрим на значения, O и M одновременно быть не могут,
# или одно, или другое

stars["Spectral Class"].value_counts()
```

```python
# хотя бы одно верно: оператор |

stars_om = stars[(stars["Spectral Class"] == "O") | 
    (stars["Spectral Class"] == "M")] 
```

```python
# ровно одно верно: оператор ^
# здесь результат | и ^ совпадают

stars_om = stars[(stars["Spectral Class"] == "O") ^ 
    (stars["Spectral Class"] == "M")] 
```

```python
# считаем среднее
stars_om["Temperature"].mean()
```

```python
# первый способ: работает в новых версиях pandas
# второй способ: работает и в более старых версиях

print(stars_om["Temperature"].mean().round(2))
print(round(stars_om["Temperature"].mean(), 2))
```

Сохраните строки, соответствующие звездам-гипергигантам и звездам-супергигантам, в датафрейм `giants`. Вычислите минимальную светимость таких звезд.

### Задача 5

Сохраните строки, соответствующие звездам главной последовательности синего (`Blue`) или сине-белого цвета (`Blue-white` или `Blue-White`), в датафрейм `main_blue`. 

Вычислите количество строк в полученном датафрейме. Выведите описательные статистики для столбца `Spectral Class`. Звезд какого спектрального класса больше всего среди отобранных наблюдений?

```python
# вариант 1: внимание на круглые скобки

main_blue = stars[(stars["Star type"] == 3) & 
      ((stars["Star color"] == "Blue") | 
       (stars["Star color"] ==  "Blue-white") | 
       (stars["Star color"] == "Blue-White"))] 
```

```python
# вариант 2: проверяем вхождение в список через .isin()
# альтернатива трем условиям с ИЛИ между ними

main_blue = stars[(stars["Star type"] == 3) & 
    (stars["Star color"].isin(["Blue", "Blue-white", "Blue-White"]))]
```

```python
# вариант 3: проверяем, входит ли строка Blue в ячейки столбца,
# через str.contains() и убираем учет регистра case = False

stars[stars["Star color"].str.contains("Blue", case = False)]
```

### Задача 6

Сохраните строки, соответствующие звездам, чья температура не ниже медианы (50-ый процентиль), но не выше верхнего квартиля (75-ый процентиль), в датафрейм `last`.

```python
# смотрим на описательные статистики

stars["Temperature"].describe()
```

```python
# отдельно вычисляем медиану med
# отдельно считаем 75-ый процентиль q
# строим фильтр, одновременно >= med и <= q

med = stars["Temperature"].median()
q = stars["Temperature"].quantile(0.75)

last = stars[(stars["Temperature"] >= med) & (stars["Temperature"] <= q)] 
```

```python
# альтернатива: вместо двух условий с &
# можно проверить вхождение в числовой интервал от .. до ..
# метод .between()

last = stars[stars["Temperature"].between(med, q)]
```

```python
# NB: если запросить help, увидим, что по умолчанию в .between()
# как раз включение обоих концов отрезка a <= x <= b,
# что нам здесь и надо

help(stars["Temperature"].between)
```
