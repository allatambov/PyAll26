### Активировать тесты

Запустите следующую ячейку, чтобы иметь возможность запускать тесты.

```python
# Фабрика тестов для проверки программ, принимающих данные через input()
# Разработчик тестов – Щуров И.В.

from collections import deque

class Tester(object):
    def __init__(self, inp):
        self.outputs = []
        self.inputs = deque(inp)
    def print(self, *args, sep = " ", end = "\n"):
        text = sep.join(map(str, args)) + end
        newlines = text.splitlines(keepends=True)
        if self.outputs and self.outputs[-1] and self.outputs[-1][-1] != "\n" and newlines:
            self.outputs[-1] += newlines[0]
            self.outputs.extend(newlines[1:])
        else:
            self.outputs.extend(newlines)
            
    def input(self, *args):
        assert self.inputs, "Вы пытаетесь считать больше элементов, чем предусмотрено условием"
        return self.inputs.popleft()
    def __enter__(self):
        global print
        global input
        print = self.print
        input = self.input
        return self.outputs
    def __exit__(self, *args):
        global print
        global input
        del print
        del input
```

### Задача 1. А не посчитать ли нам, уважаемые кроты...

Крот хочет вести учет количества рисовых зерен, которые съедает Дюймовочка. Напишите программу, которая:

* запрашивает с клавиатуры строку с дробными числами через пробел (одно значение – часть рисового зернышка, которое съедает Дюймовочка за день)
* выводит на экран суммарное число зерен с точностью до первого знака после точки.

**Примеры работы программы:**

Ввод:

    0.5 0.6 0.4 0.2 0.3 0.5

Вывод:

    2.5

Ввод:

    0.2 0.5 0.5 0.3

Вывод:

    1.5

```python
# решение 1: с созданием списка чисел

def greedy_mole():

    # список строк
    zerna_s = input().split()

    # создаем и заполняем список чисел
    zerna_num = []
    for zerno in zerna_s:
        zerna_num.append(float(zerno))

    # считаем сумму элементов и округляем
    print(round(sum(zerna_num), 1))
            
greedy_mole()
```

```python
# решение 2: без создания списка чисел

def greedy_mole():

    # список строк
    zerna_s = input().split()

    # создаем счетчик и прибавляем к нему числа
    summa = 0
    for zerno in zerna_s:
        summa += float(zerno)

    # считаем сумму элементов и округляем
    print(round(summa, 1))
            
greedy_mole()
```

```python
test_data = [
    ("0.5 0.6 0.4 0.2 0.3 0.5", "2.5"),
    ("0.2 0.5 0.5 0.3", "1.5"),
    ("0.3 0.3 0.4 0.8 0.6 0.5", "2.9")
]

for inp, out in test_data:
    with Tester([inp]) as t:
        greedy_mole()
        assert len(t) == 1, "Вам нужно вывести ровно одну строку с ответом"
        assert t[0] == out+"\n", "Неверный ответ, была введена строка " + inp
```

###  Задача 2. Jingle Spells

В списке `spells` хранятся кортежи, описывающие заклинания: заклятие (строка), контрзаклятие (строка), номер главы в учебнике, где эти заклинания описываются (целое число).

Напишите программу, которая запрашивает с клавиатуры заклинание и:

* если заклятие в списке есть, выводит на экран пару соответствующее ему контрзаклятие и через пробел номер главы, где об этом можно почитать;
* если такого заклятия нет, выводит на экран сообщение `No such spell`.

**Примеры работы программы**

Ввод:

    lumos

Вывод:

    nox 1

Ввод:

    depulso

Вывод:

    accio 20

Ввод: 

    alohomora

Вывод:

    No such spell

```python
spells = [(('lumos', 'nox'), 1), 
          (('engorgio', 'reducio'), 15), 
          (('accio', 'depulso'), 20),
          (('incarcero', 'relashio'), 5),
          (('levicorpus', 'liberacorpus'), 30)]

def jingle_spells():

    # i = 0 - счетчик для того, чтобы после выхода из цикла 
    # проверить, был ли запрос search в списке spells
    
    search = input()
    i = 0
    for pair, chapter in spells:
        if search == pair[0]:
            print(pair[1], chapter)
            i += 1
        if search == pair[1]:
            print(pair[0], chapter)
            i += 1
    if i == 0:
        print("No such spell")

jingle_spells()
```

```python
test_data = [
    ("liberacorpus", "levicorpus 30"),
    ("accio", "depulso 20"),
    ("reducio", "engorgio 15"),
    ("nox", "lumos 1")
]

for inp, out in test_data:
    with Tester([inp]) as t:
        jingle_spells()
        assert len(t) == 1, "Неверный ответ"
        assert t[0] == out+"\n", "Неверный ответ"
```

### Задача 3. Явка

В файле `elections.txt` (скачайте [здесь](https://github.com/allatambov/PyAll26/blob/main/elections.txt) и загрузите рядом с текущим файлом через *Upload*) – три строки с результатами выборов в разрезе территориальных избирательных комиссий (ТИКов):

* первая строка – название ТИК через точку с запятой
* вторая первая строка – значения явки в абсолютных числах, число полученных бюллетеней через точку с запятой
* третья строка –  общее число зарегистрированных избирателей через точку с запятой.

a. Напишите программу, которая создаст список `turnout_perc` со значениями явки на каждом ТИКе в процентах (с точностью до сотых).

b. Напишите программу, которая вычислит среднюю явку в процентах на основе `turnout_perc`, создаст список в названиями ТИКов, где явка выше средней и выведет количество таких ТИКов.

```python
# lines – список текстовых строк

lines = open("elections.txt").readlines()

# в каждой строке (0, 1, 2) убираем отступ \n на конце 
# через .strip(), разбиваем текст и получаем список текстовых значений

tiks = lines[0].strip().split(";")
turnout_s = lines[1].strip().split(";")
total_s = lines[2].strip().split(";")

# смотрим на полученные списки

print(tiks)
print(turnout_s)
print(total_s)
```

```python
# a
turnout_perc = []
for yavka, tot in zip(turnout_s, total_s):
    perc = round(int(yavka) / int(tot) * 100, 2)
    turnout_perc.append(perc)
print(turnout_perc)
```

```python
# b
avg = sum(turnout_perc) / len(turnout_perc)
tiki = []

for tik, yavka in zip(tiks, turnout_perc):
    if yavka > avg:
        tiki.append(tik)
print(len(tiki))
```
