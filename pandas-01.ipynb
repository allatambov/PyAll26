### Описание датафрейма: статистическое описание с `.describe()`

Для описания данных в статистическом смысле пригодится метод `.describe()`:

```python
df.describe()
```

По умолчанию метод выбирает числовые столбцы типов `float` и `int` и возвращает для них следующие описательные статистики:

* `count` – число заполненных ячеек, 
* `mean` – среднее арифметическое, 
* `std` – стандартное отклонение, 
* `min` и `max` – минимум и максимум, 
* `50%` – медиана, 
* `25%` – 25-ый процентиль, он же нижний квартиль, он же первый квартиль, он же квантиль уровня 0.25, 
* `75%` – 75-ый процентиль, он же верхний квартиль, он же третий квартиль, он же квантиль уровня 0.75.

Простыми словами, если мы рассмотрим описание `diffc` (`Unnamed: 0` описывать бессмысленно, это номер строки, мы его удалим позже), получается, что средняя сложность статьи равна 2.69, но при этом в 50% случаев сложность статьи не превышает 2.4. Сложность 25% статей не выше 1.9, а сложность 75% статей не выше 3.2. При этом минимальное значение сложности статьи равно 1.0, а максимальное – 9.1. Такое соотношение описательных статистик намекает на то, что в данных присутствуют новости с нетипично высокой сложностью: у 75% новостей `diffc` не выше 3.2, а максимум при этом 9.1, разрыв между этими значениями большой.

Значение стандартного отклонения оценить в данном случае сложно, оно отвечает за разброс значений относительно среднего. Чем оно выше, тем более разнообразные значения присутствуют в данных, потому что если все значения одинаковы и равны числу $a$, среднее тоже равно числу $a$, а стандартное отклонение – 0. Стандартное отклонение имеет смысл интерпретировать при сравнении групп, например, если мы захотим сравнивать разнообразие уровней сложности новостей разных рубрик (где выше – там выше разнообразие).

Как мы уже заметили, в данных преобладают текстовые столбцы, и их тоже неплохо бы описать. Добавим в `.describe()` аргумент `include` и впишем туда название типа (`str` или `object`):

```python
# в более старой версии pandas 
# df.describe(include = "object")

df.describe(include = str)
```

Здесь `count` – число заполненных ячеек, `unique` – число уникальных значений (категорий), `top` – мода, самое частое значение, `freq` – частота, соответствующая моде. Так, например, новости были предложены и написаны 22 разными авторами, чаще всего – Олегом Лищуком (954	раза). 

> Если все значения в текстовом столбце уникальны, то они все встречаются по одному разу, поэтому формально каждое из них является модой. В `top` в таком случае может находится первое значение в столбце или иное значение (от системы к системе разное, но на одном и том же компьютере одинаковое при каждом вызове `.describe()`), поскольку метод выбирает его произвольно, но не совсем случайно. Чтобы понять эту загадочную фразу, представьте, что Python не задействует алгоритм случайного выбора элемента, который каждый раз может вернуть разный результат, а просто выбирает первый из неупорядоченного множества. На одном компьютере это множество будет выглядеть одним образом, на другом – другим, и так далее.

Если вернуться к описанию числовых данных, то основные описательные статистики можно получить, применяя методы к отдельным столбцам:

```python
# выбираем столбец diffc
# и описываем

df["diffc"].describe()
```

```python
# минимум, максимум
print(df["diffc"].min())
print(df["diffc"].max())

# среднее и стандартное отклонение
print(df["diffc"].mean())
print(df["diffc"].std())
```

```python
# нижний квартиль - квантиль уровня 0.25
# нижний и верхний квартиль - квантили уровней 0.25 и 0.75

print(df["diffc"].quantile(0.25))
print(df["diffc"].quantile([0.25, 0.75]))
```

### Удаление столбцов и аргумент `inplace`

Удалим неинформативный столбец `Unnamed: 0`. Для удаления строк и столбцов на датафреймах определен универсальный метод `.drop()`. Впишем в аргумент `columns` список из названий столбцов, которые подлежат удалению:

```python
df.drop(columns = ["Unnamed: 0"])
```

Что сделал метод `.drop()`? Он вернул копию датафрейма `df`, в которой указанный столбец был удален. Сам оригинальный датафрейм `df` при этом не изменился:

```python
df
```

Как сохранить изменения? Добавить в `.drop()` аргумент `inplace = True`, который сохранит изменения в самом датафрейме, то есть перезапишет датафрейм с изменениями *на место* (отсюда и *in place*) старого:

```python
# теперь удалилось

df.drop(columns = ["Unnamed: 0"], inplace = True)
df
```

**Примечание 1.** Если вариант с `inplace` кажется загадочным, можно просто перезаписать результат `.drop()` через `=`:

    df = df.drop(columns = ["Unnamed: 0"])
    
**Примечание 2.** Если Python в ответ на `.drop()` выдает ошибку `KeyError`, это может происходить по трем причинам:

* забыли указать название аргумента `columns` и написали, например, просто `df.drop(["Unnamed: 0"])`; метод по умолчанию работает со строками, поэтому название `Unnamed: 0` Python ищет среди индексов строк, не находит такого и выдает ошибку ключа;
* указали название столбца с опечаткой;
* пытаетесь повторно удалить столбец, запустив ячейку с `.drop()` второй раз; посмотрите на датафрейм и удостоверьтесь, что столбец, который удаляете, не был удален в предыдущей операции.

### Изменение типов столбцов и добавление новых столбцов

Для начала выберем те столбцы, которые нам понадобятся для дальнейшей работы – перечислим их названия в виде списка:

```python
# внешние квадратные скобки – фильтрация, выбор столбцов
# внутренние – границы списка названия

small = df[["title", "author", "date", "diffc", "rubrics"]]
small
```

Скорректируем тип столбца `date`. Сейчас, как мы видели в `.info()`, столбец с датой – текстовый. Чтобы получить возможность забирать из даты разные элементы (год, месяц, день, день недели и прочее), а также сортировать строки правильным образом в хронологическом порядке, нужно привести столбец к типу *дата-время*. Для этого воспользуемся функцией `to_datetime()`:

```python
# перезаписываем столбец, как запись в словаре,
# старый вытесняется новым

small["date"] = pd.to_datetime(small["date"])

# проверим, что теперь тип корректный
small.info()
```

К обработке дат мы еще вернемся. 

Рассмотрим пример добавления нового столбца на основе старого. Также как и на массивах, операции на столбцах датафрейма векторизованы, достаточно описать действие, и оно будет применено ко всем ячейкам в столбце безо всяких циклов.

Добавим столбец `diffc_rel` с относительной сложностью новости – относительно максимальной сложности в процентах:

```python
#  делим на максимум по столбцу
# умножаем на 100

small["diffc_rel"] = small["diffc"] / small["diffc"].max() * 100
small
```

Теперь добавим столбец с длиной заголовка – количеством символов в тексте. Для вычисления длины текстовой строки используется функция `len()`. Просто применить ее к столбцу нельзя, `len(df["title"])` вернет количество ячеек в столбце. Поэтому функцию `len()` нужно поместить в метод `.apply()`, он применит ее отдельно к каждой ячейке с текстом:

```python
small["title_len"] = small["title"].apply(len)
small
```

> По сути `.apply()` – более эффективный аналог цикла `for` или базовой функции `map()`.

Вместо `len()` можно поставить в `.apply()` любую функцию, например, ту, которую мы предварительно напишем сами. Напишем функцию `convert(x)` для кодирования сложности новости:

* сложность 8 и выше – это 5;
* сложность от 6 до 8 – это 4;
* сложность от 4 до 6 – это 3;
* сложность ниже 4 – это 2.

```python
# ветка с x > 0  и else для страховки,
# чтобы при наличии пропусков
# применение функции не остановилось

def convert(x):
    if x >= 8:
        y = 5
    elif x >= 6:
        y = 4
    elif x >= 4:
        y = 3
    elif x >= 0:
        y = 2
    else:
        y = x
    return y

# применяем к каждой ячейке

small["diffc_cat"] = small["diffc"].apply(convert)
small
```

Если функция небольшая, в одну строчку, ее можно записать в виде lambda-функции. Например, чтобы получить количество рубрик, нам нужно строку с ними разбить по запятой с пробелом и посчитать число элементов в полученном списке:

    lambda x: len(x.split(", "))

Поместим эту функцию в `.apply()`:

```python
small["n_rubrics"] = small["rubrics"].apply(lambda x: len(x.split(", ")))
small
```

Часто в анализе данных возникает необходимость создавать **бинарные столбцы** – столбцы из 0 и 1, где 1 соответствует выполнению некоторого условия. Для простого решения этой задачи достаточно вспомнить, что для Python логические значения `True` и `False` эквивалентны 1 и 0. А значит, можно просто проверить выполнение условия для всех ячеек, получить результат в виде столбца из `True` и `False` и сменить его тип на `integer`. 

Добавим бинарный столбец `diffc_bin`, где 1 соответствует значениям сложности новости не ниже среднего:

```python
# small["diffc"] >= small["diffc"].mean() состоит из True/False,
# через .astype() меняем тип

small["diffc_bin"] = (small["diffc"] >= small["diffc"].mean()).astype(int)
small
```

В заключение вернемся к датам. Раз столбец `date` имеет тип `datetime`, из него можно извлечь отдельные элементы даты. Обычно они хранятся внутри атрибутов с говорящими названиями (`.year`, `.month` и др) внутри аксессора`dt`. Можно относится к нему как к подмодулю, который не нужно импортировать, аксессор обеспечивает доступ к набору узкоспециализированных методов и атрибутов, в данном случае для обработки дат и времени.

Извлечем год, месяц и день и сохраним их в отдельные столбцы:

```python
small["year"] = small["date"].dt.year
small["month"] = small["date"].dt.month
small["day"] = small["date"].dt.day
small
```

Чтобы забрать день недели, потребуется не атрибут, а метод – метод  `.day_name()`:

```python
small["weekday"] = small["date"].dt.day_name()
small
```

В заключение посмотрим, как рубрики новости разнести на разные столбцы. По аналогии с аксессором `dt` в `pandas` есть аксессор `str` для работы со строками. Он предоставляет доступ к методам на строках в `pandas`, а эти методы, в свою очередь, представляют собой расширенные и доработанные методы на строках в обычном Python (все те же `.split()`, `.join()`, `.upper()` и прочие).

Рассмотрим метод `.split()` внутри `str` и разобьем рубрики по запятой:

```python
small["rubrics"].str.split(",")
```

Этот вариант .`split()` применился ко всем ячейкам в столбце, теперь внутри каждой ячейки текст со списком. Чтобы «растянуть» элементы списка на отдельные столбцы, достаточно добавить аргумент `expand = True`:

```python
small2 = small["rubrics"].str.split(",", expand = True)
small2
```

Получили в `small2` датафрейм из четырех столбцов, где в каждом столбце – одна рубрика (если рубрик меньше четырех, там пропуски). Объединим два датафрейма `small` и `small2` в один – воспользуемся функцией `concat()`:

```python
final = pd.concat([small, small2], axis = 1)
final
```

Аргумент `axis = 1` гарантирует склеивание по столбцам – `small2` доклеивается справа от `small`. По умолчанию `axis = 0`, поэтому если бы мы убрали этот аргумент совсем, склеивание произошло бы по строкам – строки из `small2` добавились бы после строк `small`.

### Переименование столбцов

После объединения датафреймов в `final` остались столбцы с названиями 0, 1, 2 и 3. Что интересно, так как в `small2` Python занумеровал эти столбцы сам, сейчас они воспринимаются им как целые числа, в то время как остальные названия текстовые:

```python
# атрибут .columns – перечень названий столбцов
# последние 0-3 без кавычек

print(final.columns)
```

Переименуем столбцы, используя метод `.rename()` – подадим ему на вход словарь соответствий *старое название-новое название*:

```python
# та же история, что и с .drop()
# добавляем inplace = True, чтобы сохранить изменения сразу в final,
# в начале обязательно columns, так как по умолчанию 
# переименовываются строки

final.rename(columns = {0 : "rub1", 1 : "rub2", 
                        2 : "rub3", 3 : "rub4"}, inplace = True)
final
```

### Фильтрация наблюдений: выбор строк по условию, операторы `&` и `|`

Прежде чем переходить к фильтрации строк, вспомним немного о булевых массивах и распространим эти знания на столбцы датафрейма. Булев массив из `True` и `False` можно было получить, сформулировав условие, истинность которого автоматически проверится для всех элементов массива. Поступим аналогичным образом со столбцом датафрейма. Выберем столбец `diffc` по названию (очень похоже на выбор значения по ключу из словаря) и проверим, превышает ли сложность новостей значение 6:

```python
final["diffc"] > 6
```

Получили булев столбец – последовательность `pandas Series` из `True` и `False`, где `True` стоит на местах значений выше 6. Если нам нужно определить число новостей, удовлетворяющих условию и при этом не требуется выбирать соответствующие строки, можно просто вычислить сумму по такому столбцу:

```python
(final["diffc"] > 6).sum()
```

Итак, 79 новостей (всего 3706) имеют сложность выше 6.

Если сами строки нам все-таки нужны, и мы хотим увидеть, что это за сложные новости, условие, которое возвращает `True` или `False` нужно поместить в фильтр – в квадратные скобки:

```python
# выбираем и сохраняем в news01
# метод .head() показывает первые 5 строк

news01 = final[final["diffc"] > 6]
news01.head()
```

Фильтрация происходит следующим образом: в `news01` Python забирает те строки, на которых условие внутри квадратных скобок вернуло значение `True`.

**Примечание.** Почему `df` в коде выше встречается более одного раза – `df[df["diffc"]...]`? Посмотрите на выражение `df[...]` как на шаблон для выбора элементов. В квадратных скобках мы обычно указываем критерии выбора. При работе со списками и строками в квадратных скобках мы указывали индексы или срезы типа `0:10`, при работе со словарями – ключи записей. А для массивов и датафреймов в квадратных скобках можно указать условия для фильтрации значений. Так как датафрейм `df` состоит из разных столбцов, условия логично формулировать для каждого столбца отдельно, а не для всего `df`. А столбец мы выбираем по названию из `df` (здесь `df["diffc"]`), потому что за рамками `df` объект `diffc` для Python не существует. Поэтому и возникает двойное `df`. 

Если мы попробуем написать что-то вроде `df["diffc" > 6]` или `df[diffc > 6]`, ничего не получится. В первом случае Python будет сравнивать слово `"diffc"` с числом 6, что вызовет ошибку `TypeError`.  А во втором Python будет искать переменную `diffc`, которую мы отдельно не создавали, что вызовет ошибку `NameError`.

Выберем строки, соответствующие новостям со сложностью выше 6 из рубрики *Физика*:

```python
# объединяем условия через &
# обратите внимание на обязательные круглые скобки вокруг частей условия
# метод .head() показывает первые 5 строк

news02 = final[(final["diffc"] > 6) & 
    (final["rubrics"] == "Физика")]
news02.head()
```

**Примечание.** Если условие стоят внутри квадратных скобок, его части можно переносить по строкам. Здесь Python не выдает ошибку синтаксиса из-за переноса после `&`, потому что он понимает, что пока квадратная скобка не закрыта, далее следует продолжение предыдущей строки.

Выберем строки, соответствующие новостям со сложностью выше 6, относящиеся к рубрикам *Медицина* или *Физика*:

```python
# обратите внимание на круглые скобки, 
# порядок действий важен, сначала проверяем через |
# а потом уже к результату добавляем еще одно через &

news03 = final[(final["diffc"] > 6) & 
    ((final["rubrics"] == "Физика") | 
     (final["rubrics"] == "Медицина"))]
news03.head()
```

**Примечание.** Почему мы используем оператор `|` (`or`), а не `&` (`and`)? Оператор `&` всегда проверяет *одновременное* выполнение условий, а оператор `|` – выполнение *хотя бы одного* условия. В ячейке в столбец `rubrics` не может одновременно стоять `Зоология` и `Экология и климат`, только что-то одно. Если новость относится к нескольким рубрикам сразу, в `rubrics` названия будут перечислены через запятую с пробелом, а это совсем другое. Запросим таблиц частот для этого столбца и убедимся в этом:

```python
# метод .value_counts()
# показывает, сколько раз значения встречаются в столбце,
# значения отсортированы от самого частого к самому редкому

df["rubrics"].value_counts()
```

Если частей условия, которые мы объединяем через `|` много, удобнее вместо этого оператора использовать метод `.isin()`, он проверяет вхождение значения в список:

```python
final[final["rubrics"].isin(["Физика", "Астрономия", "Геология"])]
```

Метод `.isin()` возвращает значение `True`, если название рубрики входит в указанный список, и `False`, если название рубрики не входит в указанный список.
