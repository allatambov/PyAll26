### Немного теории: чтение и запись текстовых файлов

Создадим файл `test.txt` – откроем его для записи с нуля с флагом `"w"` (откроем соединение с файлом):

```python
myfile = open("test.txt", "w")
```

Добавим в него записи – строки с текстом:

```python
# метод .write() записывает строку с текстом в файл,
# не добавляя при этом отступы и не переходя на новую строку,
# поэтому здесь мы сами добавили \n в нужных местах

myfile.write("THIS IS A TEST\nWAIT\n")
```

Закроем файл (закроем соединение с файлом) – сохраним изменения:

```python
myfile.close()
```

Если не хочется каждый раз писать `.close()` по окончании работы, можно использовать конструкцию `with ... as ...`. Чаще всего эта конструкция и используется на практике:

```python
with open("test.txt", "w") as myfile:
    myfile.write("THIS IS A TEST\nWAIT\n")
print("Получилось то же самое!")
```

> Здесь строка `with open("test.txt", "w") as myfile` эквивалентна записи `myfile = open("test.txt", "w")`, а выход из тела конструкции (когда блок кода с отступами заканчивается) эквивалентен записи `myfile.close()`.

Откроем файл в режиме чтения и дозаписи (флаг `"r+"`). После применения метода `.read()`, файл будет прочитан с самого начала, и мы увидим текст, сохраненный в файле:

```python
with open("test.txt", "r+") as myfile:
    print(myfile.read())
```

Благодаря `.read()` курсор, который «путешествует» по файлу, остановился после последней строки с текстом. А значит, мы можем дописать что-то в конец файла и сохранить изменения:

```python
# метод .writelines() записывает в файл список строк,
# если на входе .writelines() список не из текстовых значений, 
# получим ошибку

with open("test.txt", "r+") as myfile:
    print(myfile.read())
    myfile.writelines(["3\n", "2\n", "1\n", "Start!\n"])
```

Если мы не напишем строку `myfile.read()`, Python предварительно не считает содержимое файла, а значит, курсор, «путешествующий» по файлу, будет стоять не в конце последней строки, а в начале первой. И тогда мы просто запишем новый текст поверх старого:

```python
# для чистоты эксперимента повторим создание файла,
# чтобы файл бы в том же состоянии, что и перед 
# запуском предыдущей ячейки

with open("test.txt", "w") as myfile:
    myfile.write("THIS IS A TEST\nWAIT\n")
```

```python
# запустим и откроем файл вручную

with open("test.txt", "r+") as myfile:
    myfile.writelines(["3\n", "2\n", "1\n", "Start!\n"])
```

Если нам не хочется «пролистывать» файл перед добавлением в него нового текста, мы можем открыть его в режиме дозаписи и чтения (флаг `"a+"`):

```python
# для чистоты эксперимента повторим создание файла

with open("test.txt", "w") as myfile:
    myfile.write("THIS IS A TEST\nWAIT\n")
```

```python
# запустим и откроем файл вручную

with open("test.txt", "a+") as myfile:
    myfile.writelines(["3\n", "2\n", "1\n", "Start!\n"])
```
