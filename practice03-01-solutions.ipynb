### Задача 1. Умный (не очень) будильник

Напишите программу, которая запрашивает с клавиатуры слово и целое число через пробел и выводит на экран это слово, написанное большими буквами с восклицательным знаком на конце, повторенное указанное число раз (каждый раз – с новой строки). Однако, если указанное число больше 10, слово повторяется только 10 раз, а далее выводится сообщение `Будильник устал!`.

**Примеры работы программы**
    
Ввод:

    Белоснежка 7

Вывод:

    БЕЛОСНЕЖКА!
    БЕЛОСНЕЖКА!
    БЕЛОСНЕЖКА!
    БЕЛОСНЕЖКА!
    БЕЛОСНЕЖКА!
    БЕЛОСНЕЖКА!
    БЕЛОСНЕЖКА!

Ввод:

    утро 20

Вывод:

    УТРО!
    УТРО!
    УТРО!
    УТРО!
    УТРО!
    УТРО!
    УТРО!
    УТРО!
    УТРО!
    УТРО!
    Будильник устал!

```python
word, count = input().split()

# пишем слово заглавными буквами и доклеиваем !
# превращаем текстовый count в число

message = word.upper() + "!"
n = int(count)

# если n больше 10, снижаем его до 10,
# чтобы сообщение больше 10 раз не выводилось

if n > 10:
    n = 10
for i in range(0, n):
    print(message)

# если n равно 10, выводим дополнительное сообщение

if n == 10:
    print("Будильник устал!")
```

```python
# вариант решения с for + break
# снижать n до 10 не надо,
# в нужном месте выходим из цикла через break,
# но часть с проверкой n после цикла остается

# важно: if + break должно быть перед print()
# иначе при n > 10 сообщение напечатается 11 раз
# (надо сначала проверять, а потом печатать, не наоборот)

word, count = input().split()
message = word.upper() + "!"
n = int(count)

for i in range(0, n):
    if i == 10:
        break
    print(message)
if n > 10:
    print("Будильник устал!")
```

```python
# вариант решения с циклом while
# так же если n больше 10, снижаем его до 10,
# но добавляем счетчик i и увеличиваем его на 1,
# пока не выведем сообщение n раз

word, count = input().split()
message = word.upper() + "!"
n = int(count)

if n > 10:
    n = 10
i = 0
while i < n:
    print(message)
    i = i + 1
if n == 10:
    print("Будильник устал!")
```

### Задача 2

Напишите программу, которая фильтрует героев на основе их места проживания:

* В переменную `heroes` считывается список, где хранятся имена героев через запятую с пробелом (строки).
* В переменную `places` считывается список, где хранятся места проживания героев через запятую с пробелом (строки). Каждый год соответствует студенту из списка `heroes`.
* Гарантируется, что будет введена информация минимум для одного героя.
* В переменную `place` считывается строка — место проживания, который нас интересует.
* Программа выводит имена героев, которые проживают в месте, указанном в `place`, каждое на новой строке.
* Если таких героев нет, то программа ничего не выводит.

**Примеры работы программы**

Ввод:

    Волк, Красная Шапочка, Бабушка, Мама
    Лес, Москва, Париж, Москва
    Москва

Результат:

    Красная Шапочка
    Мама

Ввод:

    Волк, Красная Шапочка, Бабушка, Айболит
    Лес, Москва, Париж, Лес
    Лес

Результат:

    Волк
    Айболит

Ввод:

    Волк, Красная Шапочка, Бабушка, Айболит
    Лес, Москва, Париж, Лес
    Лондон

Результат:

    [ничего не выводится]

```python
heroes = input().split(", ")
places = input().split(", ")
place = input()

for hero, p in zip(heroes, places):
    if p == place:
        print(hero)
```
