В файле `languages.csv` сохранены данные по вымирающим/вымершим языкам и языкам на грани исчезновения. Одна строка таблицы соответствует одному языку. 

Показатели в файле:

* `name_en`: название языка на английском;
* `countries`: страны, на которых говорят на данном языке;
* `status`: статус языка – степень угрозы существованию языка согласно ЮНЕСКО, см. ниже;
* `speakers`: число говорящих на языке;
* `location`: описание ареала языка (где распространен);
* `lat` и `lon` : широта и долгота для `location`.

Пояснения для значений в `status`:

* 1: `Vulnerable`, большинство детей говорят на языке, но чаще всего в ограниченных сферах деятельности (например, дома); 
* 2: `Definitely endangered`, дети больше не изучают язык как родной дома; 
* 3: `Severely endangered`, на языке говорит старшее поколение (бабушки/дедушки), родители могут язык понимать, но не говорить на нем со своими детьми; 
* 4: `Critically endangered`, самые молодые носители языка - старшее поколение (бабушки/дедушки и старше), но они говорят на языке редко;
* 5: `Extinct`, говорящих на языке не осталось.

Ниже написан код, который загружает данные из файла по ссылке, забирает из датафрейма столбцы с названиями `status` и `speakers`, а из столбцов – массивы значений.

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/allatambov/PyAll26/refs/heads/main/languages.csv")

speakers = df["speakers"].values
status = df["status"].values

print(speakers)
print(status)
```

Итак, `speakers` и `status` – **обычные массивы NumPy**, однако в них много значений, поэтому Python выше вывел их с `...`. Используя эти массивы (исходный датафрейм не трогаем), выполните следующие задания.

### Задача 1

На основе массива `speakers` определите, сколько в файле языков с числом говорящих менее 25 тысяч. 

**Важно.** Используйте булев массив. Выбирать значения, удовлетворяющие условию, нельзя.

```python
# создаем булев массив, где 1 – подходящее число в speakers
# считаем сумму по такому массиву – число 1

print((speakers < 25000).sum())
```

### Задача 2

На основе массива `status` определите долю языков со статусом `Critically endangered`.

**Важно.** Используйте булев массив. Выбирать значения, удовлетворяющие условию, нельзя.

```python
# создаем булев массив, где 1 – подходящее число в speakers
# считаем среднее по такому массиву = долю 1

print((status == 4).mean())
```

### Задача 3

На основе массива `status` определите долю языков со статусом `Critically endangered` или `Extinct`.

**Важно.** Используйте булев массив. Выбирать значения, удовлетворяющие условию, нельзя.

```python
# условие ИЛИ

print(((status == 4) | (status == 5)).mean())
```

### Задача 4

Сохраните в массив `chosen_zero` значения из массива `status`, которые соответствуют языкам со статусом `Vulnerable`.

```python
# status == 1 – набор из True и False
# status[status == 1] – фильтр, отбор того, где True

chosen_zero = status[status == 1]
print(len(chosen_zero)) # длина для сверки
```

### Задача 5

Сохраните в массив `chosen_one` значения из массива `speakers`, которые соответствуют языкам с числом носителей от 10 тысяч до 20 тысяч включительно.

```python
# от и до – условие с И

chosen_one = speakers[(speakers >= 10000) & (speakers <= 20000)]
print(len(chosen_one)) # длина для сверки
```

### Задача 6

Сохраните в массив `chosen_two` значения из массива `speakers`, которые соответствуют языкам с ненулевым числом носителей не более 5 тысяч.

```python
chosen_two = speakers[(speakers != 0) & (speakers <= 5000)]
print(len(chosen_two)) # длина для сверки
```

### Задача 7

Сохраните в массив `chosen_three` значения из массива `status`, которые соответствуют языкам со статусом, включающим слово `endangered` (см описание).

```python
# по описанию это коды 2, 3, 4

chosen_three = status[(status >= 2) & (status <= 4)]
print(len(chosen_three)) # длина для сверки
```
